### Find the mean, median, and mode of the feature "Age."

```python
# mean
age_column = Credit.get("Age") #From pandas doc on get
total = 0
for i in range(400):
    total += age_column.get(i)
mean = total/400
mean
```

```python
# median
age_column_cpy = age_column.copy()
age = age_column_cpy.values.copy()
age.sort()
median = (age[199]+age[200])/2
median
```

```python
# mode
age_map = dict()
for i in range(400):
    if age_column.get(i) not in age_map:
        age_map[age_column.get(i)] = 1
    else:
        age_map[age_column.get(i)] += 1
max_key = None
max_count = 0
for key in age_map.keys():
    if age_map[key] > max_count:
        max_key = age_map[key]
        max_key = key
max_key
```

### Create a subset of Credit that only contains the columns Age, Region, and Income. Call it "Subset."

```python
Subset = Credit.get(["Age", "Region", "Income"])
```

### From this subset, create two partitions: 
- One that contains records for Ages below 50, and one that contains Ages 50+
- Call them "Younger" and "Older"

```python
Younger = Subset[Subset["Age"] < 50] #from pandas getting started (titanic example)
```

```python
Older = Subset[Subset["Age"] >= 50] #from pandas getting started (titanic example)
```

### Use these partitions to compare the mean and median Income for each age group (less than 50 vs. 50+)

```python
# mean of younger
Younger["Income"].mean()
```

```python
# mean of older
Older["Income"].mean()
```

```python
# median of younger
Younger["Income"].median()
```

```python
# median of older
Older["Income"].median()
```

### What does the difference between mean and median suggest about the distribution of income?

Income for both the older age group and the younger age group is skewed right. At least fifty percent of the sampled population is below average in earnings for their respective age group.
